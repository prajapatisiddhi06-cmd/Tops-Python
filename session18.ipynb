```python
import re

text = "Call +91-9876543210 or +91-8765432109"

numbers = re.findall(r'\+91-\d{10}',text)
print(numbers)
```

```python
import re

def check_date(text):
    return bool(re.search(r'\b\d{2}/\d{2}/\d{4}\b',text))

print(check_date("Date is 25/10/2026"))
print(check_date("No date available"))
```

```python
import re
text = "Pizza Rs. 299, Burger Rs. 150, Fries Rs. 100"

prices = re.findall(r'Rs\.\s*(\d+)',text)
prices = [int(price) for price in prices]
print("Prices:",prices)
print("Total:",sum(prices))
```

```python
import re

text = "Contact abc@gmail.com or xyz@yahoo.com"
result = re.sub(r'\b[\w.-]+@[\w.-]+\.\w+\b','[hidden email]',text)
print(result)


# re.sub -> for hidding somthing
```

```python
import re

with open("comments.txt","r") as file:
    text = file.read()

usernames = re.findall(r'@[A-Za-z0-9_]{2,}',text)
unique_usernames = sorted(set(usernames))
print(unique_usernames)
```
