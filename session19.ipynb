Define a Python class called Song with attributes title, artist, and duration (in seconds). Create an object for your favorite song and print its details.

```python
class Song:
    def __init__(self,title,artist,duration):
        self.title = title
        self.artist = artist
        self.duration = duration

song1 = Song("Perfect","Ed Sheeran",263)

print("Title:",song1.title)
print("Artist:",song1.artist)
print("Duration:",song1.duration,"seconds")
```

4.
Extend the FoodOrder class by adding a method add_item(self, item_name, item_price) that adds the item to the items list and updates total_price. Demonstrate by adding two items to your order and printing the updated total.

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price

    def add_item(self, item_name, item_price):
        self.items.append(item_name)
        self.total_price += item_price

order1 = FoodOrder("Dominos", ["Pizza"], 300)

order1.add_item("Garlic Bread", 100)
order1.add_item("Coke", 50)

print("Restaurant:", order1.restaurant_name)
print("Items:", order1.items)
print("Updated Total: ₹", order1.total_price)
```
